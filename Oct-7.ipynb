```python
#coverting single series into DataFrame
s = pd.Series(
    [1,2,3,4],
    name='Number')
df = s.to_frame()
df.columns=["digits"] #This the method for renaming the columns name
print(df)
```

```python
#coverting multiple series by using dictionary method
s2 =pd.Series([9,6,5,8])
s3 = pd.Series([12,13,14,15])
d= {
"marks-A" :s,
"marks-B": s2,
"marks-c":s3}
df = pd.DataFrame(d)
df.columns=["Telugu","Maths","English"] #rename
print(df)
```

```python
#Create a DataFrame using a Dictionary --- Create a DataFrame using a dictionary with columns product,category,price,and Quantity for 5 products
import pandas as pd

products = {
    "Product": ["Laptop", "Mobile", "Tablet", "Keyboard", "Mouse"],
    "Category": ["Electronics", "Electronics", "Electronics", "Accessories", "Accessories"],
    "Price": [50000, 25000, 20000, 1500, 800],
    "Quantity": [5, 10, 7, 15, 20]
}

df = pd.DataFrame(products)

print(df)
```

```python
#Create separate Series for Name, Age, and City and combine them into one DataFrame containing 5 records

import pandas as pd

name = pd.Series(["Sharon", "Anuj", "Babu", "Charan", "Danny"])

age = pd.Series([22, 24, 23, 25, 26])

city = pd.Series(["Vizag", "Hyderabad", "Chennai", "Bangalore", "Pune"])

df = pd.DataFrame({
    "Name": name,
    "Age": age,
    "City": city
})

print(df)
```
