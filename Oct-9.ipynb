```python
import pandas as pd
employees = {
    "EmpID": [101,102,103,104],
    "Name": ["Asha",None,"Meena","Sai"],
    "Department": ["Sales","IT","HR",None],
    "Salary": [45000,6000,None,89000]
}
I =[901,902,903,904]
df_emp = pd.DataFrame(employees,index = I)
print(df_emp)
```

```python
df_emp.isnull()
```

```python
df_emp.isna()
```

```python
df_emp.isna().sum()
```

```python
df_emp.fillna({
    'Name': 'Unkown',
    'Salary': 0,
    'Department': 0
})
print(df_emp)
```
