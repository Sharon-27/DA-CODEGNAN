```python
#creating the series (Manipulating the Data)

import pandas as pd
sales = pd.Series(
    [5000, 7000, None, 9000, 8000, 10000, None],
    index=["Monday", "Tuesday", "Wednesday", "Thursday","Friday", "Saturday", "Sunday"])
print(sales)

```

```python
#Combining of two series by using Concat method

s=pd.Series([1,2,3,4,5]) #index=[9,7,8,0])
s1=pd.Series([101,102,103,104,105]) #index=[901,902,902,904,905])
print(s)
print(s1)
d=pd.concat([s,s1]) #wrong index
c=pd.concat([s,s1],ignore_index=True) #correct index
print(d)
print(c)
```

```python
#creating a data frame

import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arun", "Sneha", "Kiran", "Anjali", "Vikram", "Meena"],
    "Age": [22,24,21,25,23,26,20,27],
    "City": ["Hyderabad", "Chennai", "Banglore", "Hyderabad",
             "Mumbai", "Chennai", "Banglore", "Mumbai"],
    "Salary": [30000,40000,28000,50000,35000,45000,25000,55000]}

df = pd.DataFrame(data,index=[101,102,103,104,105,106,107,108])
print(df)
```

```python
df['Bonus'] = df['Salary'] *0.10
print(df)
```

```python
df['New Salary'] = df['Salary'] + df['Bonus']
print(df)
```

```python
df["Age"]=df["Age"]+1
print(df)
```

```python
df["New Salary"]=df["New Salary"]+500
print(df)
```

```python
df.loc[df['Name'] == 'Priya', 'City'] = 'Pune'
print(df)
```

```python
df.loc[df['City'] == 'Banglore', 'City'] = 'Hyderabad'
print(df)
```

```python
df.loc[df['City'] == 'Mumbai','Salary'] =df['Salary']*1.05
print(df)
```

```python
df.loc[df['Salary']<45000 ,'Salary'] =60000
print(df)
```
