```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/california_housing_test.csv")
```

```python
df
```

```python
df = pd.DataFrame(df)
print(df)
```

```python
#Data Frame

import pandas as pd

housing_df = pd.DataFrame({

    "longitude": [-122.05, -118.30, -117.81, -118.36, -119.67],

    "latitude": [37.37, 34.26, 33.78, 33.82, 36.33],

    "housing_median_age": [27, 43, 27, 28, 19],

    "total_rooms": [3885, 1510, 3589, 67, 1241],

    "total_bedrooms": [661, 310, 507, 15, 244],

    "population": [1537, 809, 1484, 49, 850],

    "households": [606, 277, 495, 11, 237],

    "median_income": [6.6085, 3.5990, 5.7934, 6.1359, 2.9375],

    "median_house_value": [344700, 176500, 270500, 330000, 81700]

})

print(housing_df)
```

```python
#Matplotlib Charts:-
#Line Chart

plt.plot(df.index, df['median_house_value'])
plt.title('Median House Value')
plt.xlabel('Record Index')
plt.ylabel('Median House Value')
plt.show()
```

```python
#Multiple line Chart:-
plt.plot(housing_df.index,
         housing_df['total_rooms'],
         marker='o',
         label='Total Rooms')

plt.plot(housing_df.index,
         housing_df['total_bedrooms'],
         marker='s',
         label='Total Bedrooms')

plt.plot(housing_df.index,
         housing_df['population'],
         marker='^',
         label='Population')

plt.title('Multiple Line Chart')
plt.xlabel('Housing Records')
plt.ylabel('Count')
plt.legend()
plt.grid()
plt.show()
```

```python
#Column Chart:-
plt.bar(housing_df.index,
        housing_df['population'])
plt.title('Column Chart - Population')
plt.xlabel('Housing Records')
plt.ylabel('Population')
plt.show()
```

```python
#Bar Chart:-

plt.bar(df.index[:10], df['median_house_value'][:10],color="pink")
plt.title('Median House Values')
plt.xlabel('Record Index')
plt.ylabel('House Value')
plt.show()
```

```python
#Clustered Chart:-
x = np.arange(len(housing_df))

width = 0.35

plt.bar(x - width/2, housing_df['total_rooms'],
        width, label='Total Rooms')

plt.bar(x + width/2, housing_df['total_bedrooms'],
        width, label='Total Bedrooms')

plt.title('Clustered Bar Chart')
plt.xlabel('Housing Records')
plt.ylabel('Count')
plt.xticks(x)
plt.legend()
plt.show()
```

```python
#Stacked Chart:-
plt.bar(housing_df.index, housing_df['total_rooms'],
        label='Total Rooms')

plt.bar(housing_df.index, housing_df['total_bedrooms'],
        bottom=housing_df['total_rooms'],
        label='Total Bedrooms')

plt.title('Stacked Bar Chart')
plt.xlabel('Housing Records')
plt.ylabel('Count')
plt.legend()
plt.show()
```

```python
#Horizontal Bar Chart:-
plt.barh(df.index[:10], df['median_income'][:10],color="purple")
plt.title('Median Income of First 10 Records')
plt.xlabel('Median Income')
plt.ylabel('Record Index')
plt.show()
```

```python
#Scatter Plot:-
plt.scatter(df['median_income'],df['median_house_value'],color="Green")
plt.title('Income vs House Value')
plt.xlabel('Median Income')
plt.ylabel('Median House Value')
plt.show()
```

```python
#Histogram Chart:-
plt.hist(df['median_house_value'], bins=20,color="Lavender")
plt.title('Distribution of House Values')
plt.xlabel('Median House Value')
plt.ylabel('Frequency')
plt.show()
```

```python
#Pie Chart:-
income_groups = pd.cut(
    df['median_income'],
    bins=[0, 2, 4, 6, 10, 20],
    labels=['Low', 'Below Average', 'Average', 'High', 'Very High']
)

income_count = income_groups.value_counts()

plt.pie(income_count,
        labels=income_count.index,
        autopct='%1.1f%%')

plt.title('Income Group Distribution')
plt.show()
```

```python
#Donut Chart:-
plt.pie(housing_df['median_house_value'],
        labels=housing_df.index,
        autopct='%1.1f%%',
        wedgeprops={'width': 0.4})
plt.title('Donut Chart - House Value')
plt.show()
```

```python
#Box Plot Chart:-
plt.boxplot(df['median_house_value'])
plt.title('House Value Outlier Detection')
plt.ylabel('Median House Value')
plt.show()
```

```python
#Area Chart:-
plt.fill_between(
    df.index[:50],
    df['median_house_value'][:50]
)

plt.title('House Value Area Chart')
plt.xlabel('Record Index')
plt.ylabel('Median House Value')

plt.show()
```

```python
#Violin Chart:-
plt.violinplot(df['median_income'])
plt.title('Median Income Distribution')
plt.ylabel('Median Income')
plt.show()
```

```python
#Heat Map Chart:-
correlation = df.corr()

plt.imshow(correlation, cmap='coolwarm')

plt.colorbar()

plt.xticks(range(len(correlation.columns)),
           correlation.columns,
           rotation=90)

plt.yticks(range(len(correlation.columns)),
           correlation.columns)

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
