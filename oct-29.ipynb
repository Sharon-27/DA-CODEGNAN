```python
#Choice based random
arr = np.array([10,20,30,40,50])
result = np.random.choice(arr)
print(result)
```

```python
#count
arr = np.array([10,20,30,40,50])
result = np.random.choice(arr,3)
print(result)
```

```python
arr = [10,20,30]
result = np.random.choice(
    arr,
    size=5,
    p=[0.2,0.5,0.3]
)
print(result)
```

```python
#This method is used to freeze the random values for upto the limit
arr = np.random.random(5) #random 0 to 1
np.random.seed(10)
print(arr)
```

```python
#Shuffle method(it's completely changes permenntaly)
arr = np.array([1,2,3,4,5])
np.random.shuffle(arr)
print(arr)
```

```python
#Permutations method(original will not get affected)
arr = np.array([10,20,30,40,50])
result = np.random.permutation(arr)
print("Original:",arr)
print("Permutation:",result)
```
