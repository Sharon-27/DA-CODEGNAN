```python
#Aggregate functions
import numpy as np
```

```python
arr = np.array([10,100,1000])
print(np.log10(arr))
```

```python
print(np.log2(arr))
```

```python
print(np.log(arr))
```

```python
arr = np.array([-10,-5,0,5,10])
print(np.abs(arr)) #it removes the negative sign
```

```python
#sorting integer with array
arr = np.sort([50,10,40,20,30])
print(np.sort(arr))
```

```python
#sorting string array
names = np.array([
    "Ravi",
    "Arun",
    "Priya",
    "Kiran"])
result = np.sort(names)
print(result)
```

```python
arr = np.array([50,10,40,30])
result = np.sort(arr)[::-1]
print(result)
```

```python
#Data Type Conversion
#for declaring data type-- dtype
#for converting data type-- astype
arr = np.array([10,20,30])
new_arr = arr.astype(float)
print(new_arr)
print(new_arr.dtype)
```

```python
arr = np.array([10,20,30])
new_arr = arr.astype(int)
print(new_arr)
print(new_arr.dtype)
```

```python
arr = np.array([10,20,30])
new_arr = arr.astype(str)
print(new_arr)
print(new_arr.dtype)
```

```python
arr = np.array(['10','20','30'])
new_arr = arr.astype(float)
print(new_arr)
print(new_arr.dtype)
```

```python
arr = np.array(['10','20','30'])
new_arr = arr.astype(int)
print(new_arr)
print(new_arr.dtype)
```

```python
#Slicing a 2D Array
arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print(arr)
```

```python
print(arr[0:2])
```

```python
print(arr[0:2,0:2])
```
